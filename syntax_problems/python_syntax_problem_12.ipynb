```python
from datetime import datetime as dt
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# NEW
import geopandas as gpd
from shapely.geometry import Polygon


roads_gdf = gpd.read_file('/home/jovyan/share/ne_10m_roads/ne_10m_roads.shp')

print(roads_gdf['type'].unique())

US_beltway = roads_gdf[ (roads_gdf['type'] == 'Beltway')&(roads_gdf['sov_a3']=='USA') ]
US_major = roads_gdf[ (roads_gdf['type'] == 'Major Highway')&(roads_gdf['sov_a3']=='USA') ]
US_secondary = roads_gdf[ (roads_gdf['type'] == 'Secondary Highway')&(roads_gdf['sov_a3']=='USA') ]

plt.figure(figsize=(6, 6))
ax = plt.subplot(111, projection=ccrs.LambertConformal(central_longitude=-88.75))

ax.set_extent([-88, -80, 24, 31], ccrs.PlateCarree())

ax.add_feature(cfeature.STATES.with_scale('10m'), edgecolor='black')
ax.add_feature(cfeature.COASTLINE.with_scale('10m'))

# We use this to add shapefile geometries.
ax.add_geometries(US_major['geometry'], ccrs.PlateCarree(), lw=2, edgecolor='blue', facecolor='None', label='Beltway')
ax.add_geometries(US_beltway['geometry'], ccrs.PlateCarree(), lw=2, edgecolor='red', facecolor='None', label='Major Highway')
ax.add_geometries(US_secondary['geometry'], ccrs.PlateCarree(), lw=1, edgecolor='orange', facecolor='None', label='Secondary Highway')

ax.legend()

plt.show()
```
