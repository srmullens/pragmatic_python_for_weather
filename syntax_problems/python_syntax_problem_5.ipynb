# Syntax Problem #4

## Objectives
1.	Practice writing and running Python code and using comments.
2.	Practice using the numpy package.

## Problem 
* Write a program that will read in annual tornado counts for 1950 to 2025 from `tornado.dat`. File is in the `data` folder. Compute the sum, average, and standard deviation using the numpy module.
* Write a docstring at the beginning of your code and include good comments throughout your code.

## Notes:
