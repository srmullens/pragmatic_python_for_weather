# Syntax Problem #4

## Objectives
1.	Practice writing and running Python code and using comments.
2.	Practice using control flow using definitions.
3.	Write appropriate docstring and comments.

## Problem 
Write a program that will convert a list of mean sea-level pressures from inches of mercury (inHg) to hectopascals (hPa). 
* Input data: 31.04, 29.96, 29.98, 28.17, 28.42 inHg
* Output the converted pressures in hPa.

## Notes:
* Double check to make sure you have the correct output for various pressures.
* Make sure a docstring is present, with a description of the code, inputs, and outputs.
* Make sure appropriate comments are present in your code.
* Make output informative so that anyone running your program understands what is being written without seeing the assignment.
