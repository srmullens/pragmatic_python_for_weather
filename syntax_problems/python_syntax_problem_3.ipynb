# Syntax Problem #3

## Objectives
1.	Practice writing and running Python code and using comments.
2.	Practice using control flow using if statements.
3.	Write appropriate docstring and comments.

## Problem 
Have the program read in the temperature from the standard input using the Python input() function. Write a program that will convert an input temperature in Fahrenheit to Celsius and Kelvin. Issue a “Heat Warning” if the temperature is greater than or equal to 35°C and a “Cold Warning” if the temperature is less than or equal to -5°C, otherwise indicate that the temperature is “Not too hot, not too cold”.
