## Objectives
1.	Practice writing and running Python code and using comments.

## Problem 
Write a Python program that reads in data from `tornadoes.dat` and `hail.dat`. Plot a line graph of the number of tornado and hail observations each year on the same plot. Make sure there are appropriate labels and legends. Note the data starts in 1950.
