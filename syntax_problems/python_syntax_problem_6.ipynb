## Objectives
1.	Practice writing and running Python code and using comments.
2.	Practice using the numpy package.

## Problem 
Write a Python program that reads in SOI data from soi_data.txt. Find the maximum and minimum values of SOI for a given year. Output the year, maximum SOI, and minimum SOI value for the year requested to standard output.

Use the following process to create your code:
1) Start with a specific year. Say, 2003.
2) Read in the data.
3) Determine max and min values for that year.
4) When that works, make the code generalizable so it can be used for any year.
5) For your final code, have the user input a year and output year, maximum SOI, and minimum SOI values to standard output.

## Notes:
* Double check to make sure you have the correct output.
* Make sure a docstring is present, with a description of the code, inputs, and outputs.
* Make sure appropriate comments are present in your code.
* Make output informative so that anyone running your program understands what is being written without seeing the assignment.
