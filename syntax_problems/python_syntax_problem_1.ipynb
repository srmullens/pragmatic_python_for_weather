# Syntax Problem #1 

Objectives
1.	Practice writing and running Python code in Jupyter Notebook.
2.	Use Python input/output, aka I/O, functions.

Problem 
1. Write a Python program that will print the following items:
    * Full Name 
    * Class at UF (Sophomore, Junior, Senior) based on number of semesters of college completed.
    * Hometown
    * Favorite Color
    * Favorite Book, Movie, Artist, or Song.

Notes:
* Make sure appropriate comments are present in your code.
* Make output informative so that anyone running your program understands what is being written without seeing the assignment.
* Name the program `syntax1_<username>.ipynb`. Download the file and upload it to Canvas.
