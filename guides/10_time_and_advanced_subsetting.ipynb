# Dates and Times in Python

Dealing with dates and times are not always the easiest thing. One of the built-in modules to Python is the `datetime` package, which has robust methods for dealing with all sorts of date related needs. This notebook will hit some of the main highlights of working with dates in Python.

For full information on the datetime module visit the documentation pages: https://docs.python.org/3/library/datetime.html

## Importing Datetime Module

The import statement is a little different from most of the imports that we do for Python modules as the name of the specific module we want to import is the same as the package itself. We then typically give it the shorter alias `dt`.

Documentation Page: https://docs.python.org/3/library/datetime.html#datetime-objects

Examples of Datetime Usage: https://docs.python.org/3/library/datetime.html#examples-of-usage-datetime

Note that you can format the output of your dates and times. These are used when creating strings from datetime objects (including f-string formatting) or when you have a string and you want to create the datetime object. The full documentation is here: https://docs.python.org/3/library/datetime.html#strftime-and-strptime-format-codes

A useful start is to find the current time. We find this by running `datetime.datetime.now()`, but since we imported the datetime file and its datetime class as `dt` we can just use `dt.now()`. 

In meteorology, we typically work with UTC time. Sometimes, our computer may return local time instead of UTC time. To make sure, we can specify the timezone to be UTC instead of our own. In Python, UTC is treated as a specific timezone. We specify it with `timezone.utc`.

If you want another local timzone, use the `zoneinfo` module.

````python
# Pull time from computer
date = dt.now()
print(date)

# Make sure we are getting UTC time.
date_utc = dt.now(timezone.utc)
print(date_utc)

# All available timezones
#zoneinfo.available_timezones()

# Change a datetime to another timezone.
#date_pt = date.now(zoneinfo.ZoneInfo("America/Los_Angeles"))
#print(date_pt)

# Change time zones
#date_est = date.astimezone(zoneinfo.ZoneInfo("America/New_York"))
#print(date_est)

# Format using f-string formatting.
#print(f'{date_est:%Y}')
````

For this, you use the `dt.strptime()` command and the formatting codes above. This is 'string pulled from time' but we can also think that we are 'stripping the time from the string.'
